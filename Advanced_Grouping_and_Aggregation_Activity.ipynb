
# Advanced Grouping and Aggregation in Pandas

### Objective:
In this activity, you will explore advanced grouping and aggregation techniques in Pandas. This hands-on experience will help you understand how different grouping levels and custom aggregation functions can reveal patterns in data.


### Import Pandas and Load Data
Run the code below to set up your environment and preview the dataset.

```python

import pandas as pd

data = {
    'City': ['New York', 'Los Angeles', 'New York', 'Chicago', 'Los Angeles', 'Chicago', 'New York', 'Chicago', 'Los Angeles'],
    'Category': ['Electronics', 'Furniture', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics', 'Clothing'],
    'Sales': [200, 150, 300, 400, 250, 300, 100, 500, 450],
    'Discount': [10, 15, 20, 0, 5, 10, 15, 0, 5],
    'Profit': [50, 30, 100, 150, 75, 50, 20, 200, 120]
}
df = pd.DataFrame(data)
print(df.head())
```


### Step 1: Multi-Level Grouping

Understand how to dissect data further by grouping on multiple criteria.

Execute the code below to group the dataset by `City` and `Category`, and calculate the total sales and profit for each group.


```python

# Group by City and Category and calculate metrics
grouped = df.groupby(['City', 'Category'])[['Sales', 'Profit']].sum()
print(grouped)
```

### Expected Output for Step 1:
| City        | Category      | Sales | Profit |
|-------------|---------------|-------|--------|
| Chicago     | Clothing      | 300   | 50     |
| Chicago     | Electronics   | 900   | 350    |
| Los Angeles | Clothing      | 450   | 120    |
| Los Angeles | Furniture     | 400   | 105    |
| New York    | Clothing      | 300   | 100    |
| New York    | Electronics   | 200   | 50     |
| New York    | Furniture     | 100   | 20     |



---
### Step 2: Custom Aggregation

Learn to customise the aggregation functions to gain more specific insights from the data.

Run the following code to calculate the average discount and total profit for each city. This custom aggregation provides tailored insights into the dataset.


```python

# Custom aggregation
aggregated = df.groupby('City').agg(
    avg_discount=('Discount', 'mean'),
    total_profit=('Profit', 'sum')
)
print(aggregated)
```

### Expected Output for Step 2:
| City        | avg_discount | total_profit |
|-------------|--------------|--------------|
| Chicago     | 3.33         | 350          |
| Los Angeles | 8.33         | 225          |
| New York    | 15.00        | 170          |
